# Análisis del balance

Con esta libreta se analizan y clasifican los movimientos obtenidos con el extractor de PDFs que se encuentra en el repositorio. Esto se hace con el fin de generar un presupuesto mensual basado en la frecuencia de los gastos e ingresos así como una normalización y clasificación general de las descripciones de los movimientos.

También se genera una gráfica del saldo para dar un contexto general del balance en la cuenta de débito que se analiza y conocer el patrón general de los gastos y ahorro.

## ¿Cómo se calcula el presupuesto mensual sugerido?

El presupuesto mensual sugerido se construye automáticamente a partir del historial de movimientos bancarios. Su objetivo no es simplemente calcular un promedio de los gastos pasados, sino estimar cuánto dinero convendría considerar cada mes para cubrir distintos tipos de egresos.

Para ello, el análisis considera principalmente:

- la **frecuencia** con la que aparece cada categoría;
- el **monto típico** cuando ocurre el gasto;
- el comportamiento de los **meses recientes**;
- y una **provisión mensual** para gastos que no ocurren todos los meses.

> **Importante:** el presupuesto es una estimación inicial basada en el comportamiento histórico. No sustituye las decisiones personales sobre metas de gasto, ahorro o inversión.

---

### 1. Frecuencia de los gastos

Para cada categoría se calcula en cuántos meses existió al menos un gasto:

$$
\mathrm{Frecuencia}
=
\frac{
\mathrm{Meses\;con\;gasto}
}{
\mathrm{Meses\;analizados}
}
$$

A partir de esta frecuencia, cada categoría se clasifica en uno de tres patrones:

| Patrón | Frecuencia | Interpretación |
|---|---:|---|
| **Recurrente** | ≥ 75% | El gasto ocurre prácticamente todos los meses |
| **Periódico** | 25% – 75% | El gasto ocurre regularmente, pero no todos los meses |
| **Ocasional** | < 25% | El gasto aparece solamente en algunos meses |

Esta distinción es importante porque no todos los gastos deben presupuestarse de la misma manera.

---

### 2. Mediana mensual

La columna **Mediana mensual** representa el gasto típico en aquellos meses donde la categoría tuvo actividad.

Se utiliza la mediana en lugar del promedio porque es menos sensible a meses extraordinariamente caros.

Por ejemplo, si los gastos mensuales de una categoría fueran:

`$2,500, $2,700, $2,800, $3,000, $15,000`

el gasto de `$15,000` elevaría considerablemente el promedio, mientras que la mediana seguiría representando mejor un mes habitual.

---

### 3. Promedio de los últimos 3 meses

La columna **Promedio 3 meses** permite observar el comportamiento reciente de cada categoría.

Esto ayuda a identificar situaciones donde el comportamiento actual puede ser diferente al histórico, por ejemplo:

- aumento reciente en alimentación;
- incremento en pagos de tarjeta;
- nuevos servicios recurrentes;
- reducción de algún gasto;
- aumento reciente del ahorro o inversión.

Este valor funciona principalmente como indicador de tendencia y no necesariamente determina por sí solo el presupuesto sugerido.

---

### 4. Provisión mensual

Los gastos que no ocurren todos los meses requieren un tratamiento diferente.

La provisión mensual se calcula como:

$$
\mathrm{Provision\;mensual}
=
\frac{
\mathrm{Gasto\;historico\;total\;de\;la\;categoria}
}{
\mathrm{Numero\;total\;de\;meses\;analizados}
}
$$

Esto permite distribuir gastos periódicos u ocasionales a lo largo del año.

#### Ejemplo

Si un seguro cuesta `$12,000` y se paga una vez al año, no sería conveniente presupuestar `$12,000` cada mes.

En cambio:

\[
\$12,000 / 12 = \$1,000
\]

Se podrían reservar aproximadamente `$1,000` cada mes para tener disponible el dinero cuando llegue el pago.

---

## ¿Cómo se obtiene el presupuesto sugerido?

El método depende del comportamiento de la categoría.

### Gastos recurrentes

Para los gastos que aparecen prácticamente todos los meses se utiliza principalmente el **monto mensual típico**, representado por la mediana.

También se compara con el comportamiento reciente para evitar que el presupuesto quede demasiado alejado de los hábitos actuales.

El resultado se redondea hacia arriba para producir cantidades más prácticas para presupuestar.

---

### Gastos periódicos y ocasionales

Cuando un gasto no ocurre todos los meses, utilizar directamente el monto del mes en que ocurre podría sobreestimar considerablemente el presupuesto.

Por ello se utiliza principalmente la **provisión mensual**, distribuyendo el costo histórico entre todos los meses analizados.

Este método es especialmente útil para categorías como:

- seguros;
- mantenimiento;
- salud;
- educación;
- reparaciones;
- gastos anuales o semestrales.

---

## Categorías con tratamiento especial

No todos los egresos de una cuenta bancaria representan consumo. Algunas categorías requieren una interpretación diferente.

### Pago de tarjeta de crédito

Los pagos de tarjeta se muestran explícitamente como **Pago TDC**.

Mientras no se analicen también los movimientos individuales de la tarjeta de crédito, este pago se considera una salida relevante dentro del presupuesto de la cuenta de débito.

Esto evita asumir que el pago corresponde necesariamente a deuda, ya que una tarjeta también puede utilizarse simplemente como medio de pago para gastos corrientes.

Si posteriormente se incorporan las transacciones de la tarjeta de crédito al análisis, será necesario evitar contar dos veces los mismos gastos.

---

### Ahorro e inversión

Las transferencias hacia instrumentos de ahorro o inversión se separan de los gastos de consumo.

El valor sugerido actualmente se basa en el comportamiento histórico, pero debe interpretarse como una referencia y no necesariamente como una meta óptima.

En una versión más avanzada del presupuesto, el ahorro puede definirse como un porcentaje objetivo del ingreso mensual:

**Ahorro objetivo** = **Ingreso mensual** x **Porcentaje objetivo de ahorro**

---

### Efectivo

Los retiros de efectivo representan dinero que salió de la cuenta, pero el estado de cuenta no permite conocer necesariamente en qué se gastó posteriormente.

Por esta razón, **Efectivo** se presenta como una categoría que requiere revisión.

El monto mostrado no debe interpretarse automáticamente como una recomendación de cuánto efectivo gastar cada mes.

---

### Otros

La categoría **Otros** contiene gastos que el clasificador automático todavía no pudo identificar con suficiente confianza.

Un monto elevado en esta categoría indica que conviene mejorar las reglas de clasificación o revisar manualmente algunos movimientos.

Por ello, **Otros** tampoco debe interpretarse directamente como una recomendación de gasto.

---

### Gastos extraordinarios

Los movimientos identificados como extraordinarios se excluyen del presupuesto mensual ordinario.

Ejemplos posibles incluyen:

- compra de un inmueble;
- gastos notariales;
- enganches;
- operaciones excepcionales de gran magnitud.

Estos movimientos siguen formando parte del historial financiero, pero incluirlos en el presupuesto mensual podría distorsionar significativamente la estimación.

---

## Cómo interpretar la tabla

Las columnas principales pueden leerse de la siguiente manera:

| Columna | Significado |
|---|---|
| **Categoría** | Tipo de gasto identificado |
| **Grupo** | Naturaleza financiera del movimiento |
| **Patrón** | Recurrente, periódico u ocasional |
| **Meses con gasto** | Número de meses donde apareció la categoría |
| **Frecuencia** | Porcentaje de meses donde ocurrió |
| **Mediana mensual** | Gasto típico cuando la categoría estuvo activa |
| **Promedio 3 meses** | Comportamiento reciente |
| **Provisión mensual** | Costo histórico distribuido entre todos los meses |
| **Presupuesto sugerido** | Primera estimación de cuánto considerar mensualmente |

---

## Limitaciones de la estimación

El presupuesto sugerido debe considerarse un **punto de partida**.

El comportamiento histórico describe lo que ocurrió en el pasado, pero un presupuesto también debe reflejar decisiones sobre el futuro.

Por ejemplo, una persona puede decidir:

- reducir gastos de entretenimiento;
- establecer un límite para compras;
- aumentar su ahorro;
- liquidar deuda más rápidamente;
- disminuir el uso de efectivo;
- cancelar suscripciones;
- crear fondos específicos para gastos periódicos.

Por esta razón, el objetivo final del análisis no es solamente predecir cuánto se gastará, sino utilizar el historial financiero para construir un presupuesto mensual más informado.

Una estructura objetivo puede expresarse como:

**Ingreso mensual = Gastos ordinarios + Pagos financieros + Ahorro e inversión + Disponible**

#Análisis Débito

En esta libreta se hace la visualización y análisis de los datos generados con el extractor de estados de cuenta

```python
from pathlib import Path
import re
import unicodedata

import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. CONFIGURACIÓN
# ============================================================

CONTENT_DIR = Path("/content")

ARCHIVOS_REQUERIDOS = {
    "movimientos": "movimientos_debito_total.csv",
    "saldos": "saldos_debito_total.csv",
    "resumenes": "resumenes_debito_total.csv",
}


# ============================================================
# 2. LOCALIZAR ARCHIVOS
# ============================================================

rutas = {}

for nombre, archivo in ARCHIVOS_REQUERIDOS.items():

    ruta = CONTENT_DIR / archivo

    if not ruta.exists():
        raise FileNotFoundError(
            f"No se encontró el archivo requerido:\n"
            f"{ruta}\n\n"
            f"Sube '{archivo}' a /content y vuelve a ejecutar la celda."
        )

    rutas[nombre] = ruta


print("Archivos detectados:")
print("-" * 60)

for nombre, ruta in rutas.items():
    print(f"{nombre:12} -> {ruta.name}")


# ============================================================
# 3. FUNCIÓN PARA NORMALIZAR COLUMNAS
# ============================================================

def normalizar_nombre_columna(nombre):

    nombre = str(nombre).strip().lower()

    nombre = unicodedata.normalize(
        "NFKD",
        nombre
    )

    nombre = "".join(
        caracter
        for caracter in nombre
        if not unicodedata.combining(caracter)
    )

    nombre = re.sub(
        r"[^a-z0-9]+",
        "_",
        nombre
    )

    return nombre.strip("_")


# ============================================================
# 4. FUNCIÓN PARA LEER CSV
# ============================================================

def leer_csv(ruta):

    df = pd.read_csv(
        ruta,
        sep="|",
        encoding="utf-8-sig",
        low_memory=False
    )

    df.columns = [
        normalizar_nombre_columna(col)
        for col in df.columns
    ]

    return df


# ============================================================
# 5. CARGAR LOS TRES CONSOLIDADOS
# ============================================================

df_movimientos_raw = leer_csv(
    rutas["movimientos"]
)

df_saldos_raw = leer_csv(
    rutas["saldos"]
)

df_resumenes_raw = leer_csv(
    rutas["resumenes"]
)


print("\nFilas cargadas:")
print("-" * 60)

print(
    f"Movimientos: {len(df_movimientos_raw):,}"
)

print(
    f"Saldos:      {len(df_saldos_raw):,}"
)

print(
    f"Resúmenes:   {len(df_resumenes_raw):,}"
)


# ============================================================
# 6. VALIDAR COLUMNAS DE MOVIMIENTOS
# ============================================================

columnas_movimientos_requeridas = {
    "fecha_oper",
    "fecha_liq",
    "descripcion",
    "cargo",
    "abono",
    "saldo_operacion",
    "saldo_liquidacion",
    "pagina",
    "archivo_origen",
}

faltantes = (
    columnas_movimientos_requeridas
    - set(df_movimientos_raw.columns)
)

if faltantes:

    raise ValueError(
        "Faltan columnas en movimientos:\n"
        f"{sorted(faltantes)}"
    )


# ============================================================
# 7. VALIDAR COLUMNAS DE SALDOS
# ============================================================

columnas_saldos_requeridas = {
    "archivo",
    "fecha_corte",
    "saldo_anterior",
    "saldo_final",
}

faltantes = (
    columnas_saldos_requeridas
    - set(df_saldos_raw.columns)
)

if faltantes:

    raise ValueError(
        "Faltan columnas en saldos:\n"
        f"{sorted(faltantes)}"
    )


# ============================================================
# 8. VALIDAR COLUMNAS DE RESÚMENES
# ============================================================

columnas_resumen_requeridas = {
    "archivo",
    "periodo_inicio",
    "periodo_fin",
    "fecha_corte",
    "saldo_anterior",
    "total_abonos",
    "total_cargos",
    "saldo_final",
    "extraccion_cuadra",
}

faltantes = (
    columnas_resumen_requeridas
    - set(df_resumenes_raw.columns)
)

if faltantes:

    raise ValueError(
        "Faltan columnas en resúmenes:\n"
        f"{sorted(faltantes)}"
    )


# ============================================================
# 9. CREAR COPIAS DE TRABAJO
# ============================================================

df_movimientos = df_movimientos_raw.copy()
df_saldos = df_saldos_raw.copy()
df_resumenes = df_resumenes_raw.copy()


# ============================================================
# 10. CONVERTIR FECHAS
# ============================================================

for columna in [
    "fecha_oper",
    "fecha_liq",
]:

    df_movimientos[columna] = pd.to_datetime(
        df_movimientos[columna],
        errors="coerce"
    )


df_saldos["fecha_corte"] = pd.to_datetime(
    df_saldos["fecha_corte"],
    errors="coerce"
)


for columna in [
    "periodo_inicio",
    "periodo_fin",
    "fecha_corte",
]:

    df_resumenes[columna] = pd.to_datetime(
        df_resumenes[columna],
        errors="coerce"
    )


# ============================================================
# 11. CONVERTIR IMPORTES
# ============================================================

def limpiar_importe(serie):

    return pd.to_numeric(
        serie
        .astype("string")
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False)
        .str.strip(),
        errors="coerce"
    )


# Movimientos
for columna in [
    "cargo",
    "abono",
    "saldo_operacion",
    "saldo_liquidacion",
]:

    df_movimientos[columna] = limpiar_importe(
        df_movimientos[columna]
    )


# Saldos
for columna in [
    "saldo_anterior",
    "saldo_final",
]:

    df_saldos[columna] = limpiar_importe(
        df_saldos[columna]
    )


# Resúmenes
for columna in [
    "saldo_anterior",
    "total_abonos",
    "total_cargos",
    "saldo_final",
    "saldo_promedio",
    "saldo_global",
]:

    if columna in df_resumenes.columns:

        df_resumenes[columna] = limpiar_importe(
            df_resumenes[columna]
        )


# ============================================================
# 12. LIMPIAR DESCRIPCIONES
# ============================================================

df_movimientos["descripcion"] = (
    df_movimientos["descripcion"]
    .astype("string")
    .fillna("")
    .str.replace(
        r"\s+",
        " ",
        regex=True
    )
    .str.strip()
)


# ============================================================
# 13. CREAR MONTO NETO DEL MOVIMIENTO
# ============================================================

df_movimientos["monto"] = (
    df_movimientos["abono"].fillna(0)
    - df_movimientos["cargo"].fillna(0)
)


# ============================================================
# 14. ORDENAR
# ============================================================

df_movimientos = (
    df_movimientos
    .sort_values(
        [
            "fecha_oper",
            "fecha_liq",
        ]
    )
    .reset_index(drop=True)
)


df_saldos = (
    df_saldos
    .sort_values("fecha_corte")
    .reset_index(drop=True)
)


df_resumenes = (
    df_resumenes
    .sort_values("fecha_corte")
    .reset_index(drop=True)
)


# ============================================================
# 15. VALIDAR EXTRACCIÓN
# ============================================================

# Puede venir como bool o como texto dependiendo de pandas.
extraccion_ok = (
    df_resumenes["extraccion_cuadra"]
    .astype("string")
    .str.lower()
    .eq("true")
)

total_estados = len(
    df_resumenes
)

estados_correctos = int(
    extraccion_ok.sum()
)

estados_con_error = (
    total_estados
    - estados_correctos
)


print("\nValidación del extractor")
print("-" * 60)

print(
    f"Estados de cuenta:    {total_estados}"
)

print(
    f"Extracción correcta:  {estados_correctos}"
)

print(
    f"Extracción con error: {estados_con_error}"
)


if estados_con_error > 0:

    print(
        "\nEstados que requieren revisión:"
    )

    display(
        df_resumenes.loc[
            ~extraccion_ok,
            [
                "archivo",
                "fecha_corte",
                "diferencia_cargos",
                "diferencia_abonos",
            ]
        ]
    )


# ============================================================
# 16. VALIDACIONES BÁSICAS
# ============================================================

fechas_invalidas = (
    df_movimientos["fecha_oper"]
    .isna()
    .sum()
)

print(
    f"\nFechas de operación inválidas: "
    f"{fechas_invalidas:,}"
)


# No eliminar automáticamente fechas futuras.
# Solo se reportan para detectar posibles errores del extractor.

fecha_actual = pd.Timestamp.today().normalize()

fechas_futuras = (
    df_movimientos["fecha_oper"]
    > fecha_actual
)

print(
    "Movimientos posteriores a hoy:",
    int(fechas_futuras.sum())
)


# ============================================================
# 17. RESUMEN GENERAL
# ============================================================

print("\nResumen de datos")
print("-" * 60)

if not df_movimientos.empty:

    print(
        f"Periodo movimientos: "
        f"{df_movimientos['fecha_oper'].min().date()} "
        f"a "
        f"{df_movimientos['fecha_oper'].max().date()}"
    )

print(
    f"Movimientos: "
    f"{len(df_movimientos):,}"
)

print(
    f"Total cargos: "
    f"${df_movimientos['cargo'].sum():,.2f}"
)

print(
    f"Total abonos: "
    f"${df_movimientos['abono'].sum():,.2f}"
)

print(
    f"Flujo neto: "
    f"${df_movimientos['monto'].sum():,.2f}"
)


if not df_saldos.empty:

    print(
        f"\nÚltimo saldo reportado por BBVA:"
    )

    ultima_fila = (
        df_saldos
        .dropna(
            subset=["fecha_corte"]
        )
        .iloc[-1]
    )

    print(
        f"Fecha de corte: "
        f"{ultima_fila['fecha_corte'].date()}"
    )

    print(
        f"Saldo final: "
        f"${ultima_fila['saldo_final']:,.2f}"
    )


# ============================================================
# 18. MOSTRAR MUESTRAS
# ============================================================

print("\nMovimientos:")
display(
    df_movimientos.head()
)

print("\nSaldos:")
display(
    df_saldos.head()
)

print("\nResúmenes:")
display(
    df_resumenes.head()
)
```

# Grafica
Ahora podemos hacer una gráfica mucho más sencilla y, sobre todo, con un significado financiero claro: cada punto será el saldo_final que BBVA reportó en el estado de cuenta de esa fecha de corte.

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import matplotlib.ticker as ticker


# ============================================================
# 1. PREPARAR DATOS
# ============================================================

df_saldo_plot = df_saldos.copy()

df_saldo_plot["fecha_corte"] = pd.to_datetime(
    df_saldo_plot["fecha_corte"],
    errors="coerce"
)

df_saldo_plot["saldo_final"] = pd.to_numeric(
    df_saldo_plot["saldo_final"],
    errors="coerce"
)

df_saldo_plot = (
    df_saldo_plot
    .dropna(
        subset=[
            "fecha_corte",
            "saldo_final"
        ]
    )
    .sort_values("fecha_corte")
    .reset_index(drop=True)
)

if df_saldo_plot.empty:
    raise ValueError(
        "No hay saldos válidos para graficar."
    )


# ============================================================
# 2. CREAR RANGO COMPLETO DE MESES
# ============================================================

fecha_inicio = (
    df_saldo_plot["fecha_corte"]
    .min()
    .to_period("M")
    .to_timestamp()
)

fecha_fin = (
    df_saldo_plot["fecha_corte"]
    .max()
    .to_period("M")
    .to_timestamp()
)

meses_completos = pd.date_range(
    start=fecha_inicio,
    end=fecha_fin,
    freq="MS"
)


# ============================================================
# 3. CREAR GRÁFICA
# ============================================================

fig, ax = plt.subplots(
    figsize=(14, 6)
)

ax.plot(
    df_saldo_plot["fecha_corte"],
    df_saldo_plot["saldo_final"],
    marker="o",
    markersize=4,
    linewidth=1.5,
    label="Saldo final reportado por BBVA"
)


# ============================================================
# 4. LÍNEA DE CERO
# ============================================================

ax.axhline(
    0,
    linestyle="--",
    alpha=0.5
)


# ============================================================
# 5. SOMBREADO ALTERNADO POR AÑO
# ============================================================

anios = range(
    df_saldo_plot["fecha_corte"].min().year,
    df_saldo_plot["fecha_corte"].max().year + 1
)

for indice, anio in enumerate(anios):

    if indice % 2 == 0:

        inicio_anio = pd.Timestamp(
            year=anio,
            month=1,
            day=1
        )

        fin_anio = pd.Timestamp(
            year=anio + 1,
            month=1,
            day=1
        )

        ax.axvspan(
            inicio_anio,
            fin_anio,
            alpha=0.08
        )


# ============================================================
# 6. EJE X — TODOS LOS MESES
# ============================================================

ax.set_xticks(
    meses_completos
)

ax.xaxis.set_major_formatter(
    mdates.DateFormatter("%Y-%m")
)

plt.setp(
    ax.get_xticklabels(),
    rotation=90,
    ha="center"
)

ax.set_xlim(
    fecha_inicio - pd.Timedelta(days=10),
    fecha_fin
    + pd.offsets.MonthEnd(1)
    + pd.Timedelta(days=10)
)


# ============================================================
# 7. EJE Y — FORMATO MONETARIO
# ============================================================

ax.yaxis.set_major_formatter(
    ticker.StrMethodFormatter("${x:,.0f}")
)


# ============================================================
# 8. MARCAR EL ÚLTIMO SALDO
# ============================================================

ultima_fila = df_saldo_plot.iloc[-1]

ultima_fecha = ultima_fila["fecha_corte"]
ultimo_saldo = ultima_fila["saldo_final"]

ax.scatter(
    ultima_fecha,
    ultimo_saldo,
    zorder=5
)

ax.annotate(
    f"${ultimo_saldo:,.2f}",
    xy=(
        ultima_fecha,
        ultimo_saldo
    ),
    xytext=(8, 8),
    textcoords="offset points",
    fontsize=9
)


# ============================================================
# 9. ESTÉTICA
# ============================================================

ax.set_title(
    "Evolución del saldo real de la cuenta de débito"
)

ax.set_xlabel(
    "Fecha de corte"
)

ax.set_ylabel(
    "Saldo final"
)

ax.grid(
    axis="y",
    alpha=0.3
)

ax.legend()

fig.tight_layout()

plt.show()


# ============================================================
# 10. RESUMEN DEL ÚLTIMO SALDO
# ============================================================

print(
    f"Último corte disponible: "
    f"{ultima_fecha.date()}"
)

print(
    f"Saldo final reportado: "
    f"${ultimo_saldo:,.2f}"
)
```

# Validación de saldos
Queremos responder por mes: cuánto entró, cuánto salió, cuál fue el flujo neto y cómo se relaciona con el saldo real de cierre.

```python
# ============================================================
# FLUJO MENSUAL DE LA CUENTA
# ============================================================

df_flujo_mensual = df_movimientos.copy()

# Usamos la fecha de operación
df_flujo_mensual["mes"] = (
    df_flujo_mensual["fecha_oper"]
    .dt.to_period("M")
)

# ============================================================
# AGRUPAR POR MES
# ============================================================

df_flujo_mensual = (
    df_flujo_mensual
    .groupby("mes")
    .agg(
        ingresos=("abono", "sum"),
        gastos=("cargo", "sum"),
        movimientos=("monto", "size")
    )
    .reset_index()
)

# Flujo neto del mes
df_flujo_mensual["flujo_neto"] = (
    df_flujo_mensual["ingresos"]
    - df_flujo_mensual["gastos"]
)

# Convertir periodo a fecha para facilitar gráficas posteriores
df_flujo_mensual["fecha_mes"] = (
    df_flujo_mensual["mes"]
    .dt.to_timestamp()
)

# Ordenar
df_flujo_mensual = (
    df_flujo_mensual
    .sort_values("fecha_mes")
    .reset_index(drop=True)
)

display(df_flujo_mensual)
```

```python
# ============================================================
# VALIDACIÓN POR PERIODO REAL DEL ESTADO DE CUENTA
# ============================================================

registros = []

for _, resumen in df_resumenes.iterrows():

    inicio = resumen["periodo_inicio"]
    fin = resumen["periodo_fin"]

    if pd.isna(inicio) or pd.isna(fin):
        continue

    # Movimientos cuya fecha de operación pertenece
    # al periodo exacto reportado por BBVA
    movimientos_periodo = df_movimientos.loc[
        (df_movimientos["fecha_oper"] >= inicio)
        &
        (df_movimientos["fecha_oper"] <= fin)
    ]

    cargos_calculados = (
        movimientos_periodo["cargo"]
        .fillna(0)
        .sum()
    )

    abonos_calculados = (
        movimientos_periodo["abono"]
        .fillna(0)
        .sum()
    )

    flujo_calculado = (
        abonos_calculados
        - cargos_calculados
    )

    cambio_saldo = (
        resumen["saldo_final"]
        - resumen["saldo_anterior"]
    )

    registros.append({
        "fecha_corte": resumen["fecha_corte"],

        "periodo_inicio": inicio,
        "periodo_fin": fin,

        "saldo_anterior": resumen["saldo_anterior"],

        "abonos_bbva": resumen["total_abonos"],
        "abonos_calculados": abonos_calculados,

        "cargos_bbva": resumen["total_cargos"],
        "cargos_calculados": cargos_calculados,

        "flujo_calculado": flujo_calculado,

        "cambio_saldo": cambio_saldo,

        "saldo_final": resumen["saldo_final"],
    })


df_validacion_periodos = pd.DataFrame(
    registros
)


# ============================================================
# DIFERENCIAS
# ============================================================

df_validacion_periodos["dif_abonos"] = (
    df_validacion_periodos["abonos_calculados"]
    - df_validacion_periodos["abonos_bbva"]
)

df_validacion_periodos["dif_cargos"] = (
    df_validacion_periodos["cargos_calculados"]
    - df_validacion_periodos["cargos_bbva"]
)

df_validacion_periodos["dif_flujo"] = (
    df_validacion_periodos["flujo_calculado"]
    - df_validacion_periodos["cambio_saldo"]
)


# ============================================================
# VALIDACIÓN
# ============================================================

TOLERANCIA = 0.02

df_validacion_periodos["cuadra"] = (
    df_validacion_periodos["dif_abonos"].abs().le(TOLERANCIA)
    &
    df_validacion_periodos["dif_cargos"].abs().le(TOLERANCIA)
    &
    df_validacion_periodos["dif_flujo"].abs().le(TOLERANCIA)
)


# ============================================================
# MOSTRAR
# ============================================================

columnas_mostrar = [
    "fecha_corte",
    "saldo_anterior",
    "abonos_bbva",
    "abonos_calculados",
    "cargos_bbva",
    "cargos_calculados",
    "flujo_calculado",
    "saldo_final",
    "dif_flujo",
    "cuadra",
]

display(
    df_validacion_periodos[
        columnas_mostrar
    ]
)
```

```python
# ============================================================
# NORMALIZACIÓN GENÉRICA DE DESCRIPCIONES
# ============================================================

df_clasificacion = df_movimientos.copy()

df_clasificacion["descripcion_original"] = (
    df_clasificacion["descripcion"]
    .astype("string")
    .fillna("")
)


# ============================================================
# 1. FUNCIÓN DE NORMALIZACIÓN
# ============================================================

def normalizar_descripcion(texto):

    texto = str(texto).upper().strip()

    # --------------------------------------------------------
    # ESPACIOS
    # --------------------------------------------------------

    texto = re.sub(
        r"\s+",
        " ",
        texto
    )


    # --------------------------------------------------------
    # RUIDO GENERADO POR EL PDF / ESTADO DE CUENTA
    # --------------------------------------------------------

    patrones_ruido = [
        r"RENDIMIENTO QUE OBTENDR[IÍ]A.*$",
        r"INSTITUCI[ÓO]N DE BANCA M[ÚU]LTIPLE.*$",
        r"RETIROS REALIZADOS:\s*\d+.*$",
        r"DEP[ÓO]SITOS REALIZADOS:\s*\d+.*$",
        r"SALDO PROMEDIO.*$",
        r"PRODUCTO PROTEGIDO.*$",
        r"REFERENCIA\s+\*+\d+.*$",
    ]

    for patron in patrones_ruido:

        texto = re.sub(
            patron,
            "",
            texto,
            flags=re.IGNORECASE
        )


    # --------------------------------------------------------
    # NORMALIZAR NOMBRES DE OPERACIONES BANCARIAS
    # --------------------------------------------------------

    reemplazos = {
        r"\bPAGO DE N[ÓO]MINA\b":
            "PAGO DE NOMINA",

        r"\bRETIRO SIN TARJETA QR\b":
            "RETIRO SIN TARJETA QR",

        r"\bPAGO TARJETA DE CR[EÉ]DITO\b":
            "PAGO TARJETA DE CREDITO",

        r"\bTRASPASO ENTRE CUENTAS\b":
            "TRASPASO ENTRE CUENTAS",
    }

    for patron, reemplazo in reemplazos.items():

        texto = re.sub(
            patron,
            reemplazo,
            texto,
            flags=re.IGNORECASE
        )


    # --------------------------------------------------------
    # PAGO CUENTA DE TERCERO
    #
    # Quitar identificadores técnicos BNET, conservando
    # el concepto escrito por el usuario.
    # --------------------------------------------------------

    texto = re.sub(
        r"^PAGO CUENTA DE TERCERO\s+BNET\s+\d+\s*",
        "",
        texto
    )


    # --------------------------------------------------------
    # SPEI
    #
    # Eliminar folios/referencias, pero conservar banco,
    # concepto o beneficiario cuando exista.
    # --------------------------------------------------------

    texto = re.sub(
        r"\bREFERENCIA\s+\d+\b",
        "",
        texto
    )

    texto = re.sub(
        r"\bMBAN[A-Z0-9]+\b",
        "",
        texto
    )

    # Cuentas CLABE largas
    texto = re.sub(
        r"\b\d{18}\b",
        "",
        texto
    )


    # --------------------------------------------------------
    # RFC
    # --------------------------------------------------------

    texto = re.sub(
        r"\bRFC[:\s]+[A-Z0-9]{10,13}\b",
        "",
        texto
    )


    # --------------------------------------------------------
    # AUTORIZACIONES
    # --------------------------------------------------------

    texto = re.sub(
        r"\bAUT[:\s]*\d+\b",
        "",
        texto
    )


    # --------------------------------------------------------
    # FOLIOS
    # --------------------------------------------------------

    texto = re.sub(
        r"\bFOLIO[:\s]*\d+\b",
        "",
        texto
    )


    # --------------------------------------------------------
    # LIMPIEZA FINAL
    # --------------------------------------------------------

    texto = re.sub(
        r"\s+",
        " ",
        texto
    )

    return texto.strip()


# ============================================================
# 2. CREAR DESCRIPCIÓN NORMALIZADA
# ============================================================

df_clasificacion["descripcion_base"] = (
    df_clasificacion[
        "descripcion_original"
    ]
    .apply(normalizar_descripcion)
)


# ============================================================
# 3. RESUMEN
# ============================================================

df_descripciones_base = (
    df_clasificacion
    .groupby(
        "descripcion_base",
        dropna=False
    )
    .agg(
        apariciones=(
            "descripcion_base",
            "size"
        ),

        primera_fecha=(
            "fecha_oper",
            "min"
        ),

        ultima_fecha=(
            "fecha_oper",
            "max"
        ),

        total_cargos=(
            "cargo",
            "sum"
        ),

        total_abonos=(
            "abono",
            "sum"
        ),
    )
    .reset_index()
)


df_descripciones_base["monto_total"] = (
    df_descripciones_base["total_cargos"]
    +
    df_descripciones_base["total_abonos"]
)


df_descripciones_base = (
    df_descripciones_base
    .sort_values(
        [
            "apariciones",
            "monto_total"
        ],
        ascending=[
            False,
            False
        ]
    )
    .reset_index(drop=True)
)


print(
    "Descripciones únicas originales:",
    df_clasificacion[
        "descripcion_original"
    ].nunique()
)

print(
    "Descripciones únicas normalizadas:",
    df_clasificacion[
        "descripcion_base"
    ].nunique()
)

display(
    df_descripciones_base.head(20)
)
```

```python
# ============================================================
# CLASIFICACIÓN GENÉRICA ORIENTADA A PRESUPUESTO
# ============================================================

df_presupuesto = df_clasificacion.copy()


# ============================================================
# 1. CATÁLOGOS DE PALABRAS CLAVE
# ============================================================

REGLAS_CATEGORIA = {

    "vivienda": [
        "HIPOTECA",
        "RENTA",
        "ARRENDAMIENTO",
        "MANTENIMIENTO",
        "MTTO",
        "IMPERMEABIL",
        "HOME DEPOT",
        "MUEBLE",
        "CLOSET",
        "SACMEX",
        "AGUA",
        "CFE",
        "LUZ",
        "GAS NATURAL",
    ],

    "alimentacion": [
        "WALMART",
        "WAL MART",
        "SAMS",
        "COSTCO",
        "CHEDRAUI",
        "SORIANA",
        "LA COMER",
        "CITY MARKET",
        "FRESKO",
        "DESPENSA",
        "SUPERMERCADO",
        "COMIDA",
        "RESTAURANTE",
        "FONDA",
        "PIZZA",
        "DOMINOS",
        "HAMBURGUESA",
        "TACOS",
        "SUSHI",
        "DESAYUNO",
        "STARBUCKS",
        "RAPPI",
        "UBER EATS",
    ],

    "transporte": [
        "UBER",
        "DIDI",
        "CABIFY",
        "GASOLINA",
        "GASOLINERA",
        "PEMEX",
        "ESTACIONAMIENTO",
        "PARKING",
        "CASETA",
        "TAG",
    ],

    "telecomunicaciones": [
        "TELMEX",
        "TELEFONOS DE MEXICO",
        "TELCEL",
        "MOVISTAR",
        "AT&T",
        "TOTALPLAY",
        "IZZI",
        "MEGACABLE",
    ],

    "salud": [
        "FARMACIA",
        "FARMACIAS",
        "MEDICO",
        "MEDICA",
        "DOCTOR",
        "CONSULTA",
        "PEDIATRA",
        "DENTISTA",
        "DENTAL",
        "LABORATORIO",
        "HOSPITAL",
        "CIRUGIA",
        "PSICO",
        "TERAPIA",
    ],

    "educacion": [
        "COLEGIATURA",
        "ESCUELA",
        "UNIVERSIDAD",
        "CURSO",
        "CLASE",
        "ACADEMIA",
        "COLEGIO",
        "INSCRIPCION",
    ],

    "seguros": [
        "GNP",
        "AXA",
        "ALLIANZ",
        "METLIFE",
        "MAPFRE",
        "QUALITAS",
        "SEGURO",
    ],

    "pago_tdc": [
        "PAGO TARJETA DE CREDITO",
        "PAGO TARJETA",
        "PAGO TDC",
    ],

    "deuda": [
        "PAGO PRESTAMO",
        "PAGO CREDITO",
        "CREDITO PERSONAL",
        "CREDITO AUTOMOTRIZ",
        "HIPOTECA",
    ],

    "ahorro_inversion": [
        "CETES",
        "CETESDIRECTO",
        "GBM",
        "ACTINVER",
        "KUSPIT",
        "FONDO DE INVERSION",
        "INVERSION",
        "NU MEXICO",
    ],

    "suscripciones": [
        "NETFLIX",
        "SPOTIFY",
        "DISNEY",
        "HBO",
        "MAX.COM",
        "YOUTUBE",
        "APPLE.COM",
        "GOOGLE",
        "AMAZON PRIME",
    ],

    "compras": [
        "AMAZON",
        "MERCADO LIBRE",
        "MERCADOLIBRE",
        "LIVERPOOL",
        "PALACIO DE HIERRO",
        "SEARS",
        "ZARA",
        "H&M",
    ],

    "entretenimiento": [
        "CINE",
        "CINEMEX",
        "CINEPOLIS",
        "BOLETOS",
        "TICKETMASTER",
        "VIAJE",
        "HOTEL",
        "STEAM",
    ],

    "efectivo": [
        "RETIRO SIN TARJETA",
        "RETIRO CAJERO",
        "RETIRO",
        "EFECTIVO",
    ],

    "comisiones": [
        "COMISION",
        "IVA COM",
        "SERV BANCA",
        "REPOSICION DE TARJETA",
        "ADMINISTRACION",
    ],
}


# ============================================================
# 2. REGLAS PARA MOVIMIENTOS EXTRAORDINARIOS
# ============================================================

PALABRAS_EXTRAORDINARIO = [
    "COMPRAVENTA",
    "COMPRA VENTA",
    "GASTOS NOTARIALES",
    "NOTARIA",
    "ESCRITURA",
    "ENGANCHE",
]


# ============================================================
# 3. HELPERS
# ============================================================

def contiene_alguna(texto, palabras):

    texto = str(texto).upper()

    return any(
        palabra in texto
        for palabra in palabras
    )


def detectar_categoria(texto):

    texto = str(texto).upper()

    for categoria, palabras in REGLAS_CATEGORIA.items():

        if contiene_alguna(
            texto,
            palabras
        ):
            return categoria

    return None


# ============================================================
# 4. CLASIFICADOR PRINCIPAL
# ============================================================

def clasificar_movimiento(fila):

    descripcion = str(
        fila["descripcion_base"]
    ).upper()

    comercio = str(
        fila.get(
            "comercio_base",
            ""
        )
    ).upper()

    texto = (
        descripcion
        + " "
        + comercio
    )

    cargo = (
        fila["cargo"]
        if pd.notna(fila["cargo"])
        else 0
    )

    abono = (
        fila["abono"]
        if pd.notna(fila["abono"])
        else 0
    )


    # --------------------------------------------------------
    # A. INGRESOS
    # --------------------------------------------------------

    if abono > 0 and cargo == 0:

        if contiene_alguna(
            texto,
            [
                "NOMINA",
                "SUELDO",
                "SALARIO",
            ]
        ):
            return pd.Series({
                "tipo_movimiento": "ingreso",
                "categoria_presupuesto": "ingreso_laboral",
                "incluir_presupuesto": False,
                "es_extraordinario": False,
            })

        if contiene_alguna(
            texto,
            [
                "PRESTAMO OTORGADO",
                "CREDITO OTORGADO",
            ]
        ):
            return pd.Series({
                "tipo_movimiento": "financiamiento",
                "categoria_presupuesto": "financiamiento",
                "incluir_presupuesto": False,
                "es_extraordinario": True,
            })

        if contiene_alguna(
            texto,
            [
                "TRASPASO ENTRE CUENTAS",
                "SPEI RECIBIDO",
                "TRANSFERENCIA RECIBIDA",
            ]
        ):
            return pd.Series({
                "tipo_movimiento": "transferencia",
                "categoria_presupuesto": "transferencia",
                "incluir_presupuesto": False,
                "es_extraordinario": False,
            })

        return pd.Series({
            "tipo_movimiento": "ingreso",
            "categoria_presupuesto": "otros_ingresos",
            "incluir_presupuesto": False,
            "es_extraordinario": False,
        })


    # --------------------------------------------------------
    # B. EGRESOS EXTRAORDINARIOS
    # --------------------------------------------------------

    if contiene_alguna(
        texto,
        PALABRAS_EXTRAORDINARIO
    ):
        return pd.Series({
            "tipo_movimiento": "gasto_extraordinario",
            "categoria_presupuesto": "extraordinarios",
            "incluir_presupuesto": False,
            "es_extraordinario": True,
        })


    # --------------------------------------------------------
    # C. CATEGORÍAS IDENTIFICABLES
    #
    # Esto se evalúa ANTES de tratar un SPEI como transferencia.
    #
    # Ejemplo:
    # SPEI ENVIADO ... CETES
    # -> ahorro_inversion
    #
    # SPEI ENVIADO ... CIRUGIA
    # -> salud
    # --------------------------------------------------------

    categoria = detectar_categoria(
      texto
    )

    if categoria is not None:

    # ----------------------------------------------------
    # PAGO DE TARJETA DE CREDITO
    # ----------------------------------------------------

        if categoria == "pago_tdc":

            return pd.Series({
                "tipo_movimiento": "transferencia",
                "categoria_presupuesto": "pago_tdc",
                "incluir_presupuesto": True,
                "es_extraordinario": False,
            })


        # ----------------------------------------------------
        # AHORRO / INVERSION
        # ----------------------------------------------------

        if categoria == "ahorro_inversion":

            return pd.Series({
                "tipo_movimiento": "ahorro_inversion",
                "categoria_presupuesto": categoria,
                "incluir_presupuesto": True,
                "es_extraordinario": False,
            })


        # ----------------------------------------------------
        # DEUDA REAL
        # ----------------------------------------------------

        if categoria == "deuda":

            return pd.Series({
                "tipo_movimiento": "deuda",
                "categoria_presupuesto": categoria,
                "incluir_presupuesto": True,
                "es_extraordinario": False,
            })


        # ----------------------------------------------------
        # RESTO DE CATEGORIAS DE GASTO
        # ----------------------------------------------------

        return pd.Series({
            "tipo_movimiento": "gasto",
            "categoria_presupuesto": categoria,
            "incluir_presupuesto": True,
            "es_extraordinario": False,
        })


    # --------------------------------------------------------
    # D. TRANSFERENCIAS SIN PROPÓSITO IDENTIFICABLE
    # --------------------------------------------------------

    if contiene_alguna(
        texto,
        [
            "SPEI ENVIADO",
            "TRANSFERENCIA ENVIADA",
            "TRASPASO ENTRE CUENTAS",
        ]
    ):

        return pd.Series({
            "tipo_movimiento": "transferencia",
            "categoria_presupuesto": "transferencia_sin_clasificar",
            "incluir_presupuesto": False,
            "es_extraordinario": False,
        })


    # --------------------------------------------------------
    # E. CARGO SIN CLASIFICAR
    # --------------------------------------------------------

    if cargo > 0:

        return pd.Series({
            "tipo_movimiento": "gasto",
            "categoria_presupuesto": "otros",
            "incluir_presupuesto": True,
            "es_extraordinario": False,
        })


    # --------------------------------------------------------
    # F. FALLBACK
    # --------------------------------------------------------

    return pd.Series({
        "tipo_movimiento": "desconocido",
        "categoria_presupuesto": "otros",
        "incluir_presupuesto": False,
        "es_extraordinario": False,
    })


# ============================================================
# 5. APLICAR CLASIFICADOR
# ============================================================

columnas_clasificacion = (
    df_presupuesto
    .apply(
        clasificar_movimiento,
        axis=1
    )
)

df_presupuesto = pd.concat(
    [
        df_presupuesto,
        columnas_clasificacion
    ],
    axis=1
)


# ============================================================
# 6. MES
# ============================================================

df_presupuesto["mes"] = (
    df_presupuesto["fecha_oper"]
    .dt.to_period("M")
)


# ============================================================
# 7. RESUMEN GENERAL DE LA CLASIFICACIÓN
# ============================================================

df_resumen_presupuesto = (
    df_presupuesto
    .groupby(
        [
            "tipo_movimiento",
            "categoria_presupuesto",
            "incluir_presupuesto",
        ]
    )
    .agg(
        movimientos=("monto", "size"),
        cargos=("cargo", "sum"),
        abonos=("abono", "sum"),
    )
    .reset_index()
)


df_resumen_presupuesto["monto_total"] = (
    df_resumen_presupuesto["cargos"]
    +
    df_resumen_presupuesto["abonos"]
)


df_resumen_presupuesto = (
    df_resumen_presupuesto
    .sort_values(
        "monto_total",
        ascending=False
    )
    .reset_index(drop=True)
)


display(
    df_resumen_presupuesto
)
```

```python
# ============================================================
# GENERAR PRESUPUESTO MENSUAL AUTOMÁTICO
# ============================================================

import numpy as np


# ============================================================
# 1. MOVIMIENTOS QUE SÍ PARTICIPAN EN EL PRESUPUESTO
# ============================================================

df_base_presupuesto = (
    df_presupuesto.loc[
        df_presupuesto["incluir_presupuesto"]
        &
        (df_presupuesto["cargo"].fillna(0) > 0)
    ]
    .copy()
)


if df_base_presupuesto.empty:
    raise ValueError(
        "No hay egresos disponibles para generar el presupuesto."
    )


# ============================================================
# 2. CREAR PERIODO MENSUAL
# ============================================================

df_base_presupuesto["mes"] = (
    df_base_presupuesto["fecha_oper"]
    .dt.to_period("M")
)


primer_mes = (
    df_base_presupuesto["mes"]
    .min()
)

ultimo_mes = (
    df_base_presupuesto["mes"]
    .max()
)


todos_los_meses = pd.period_range(
    start=primer_mes,
    end=ultimo_mes,
    freq="M"
)


total_meses = len(
    todos_los_meses
)


print(
    f"Periodo usado para presupuesto: "
    f"{primer_mes} a {ultimo_mes}"
)

print(
    f"Meses analizados: "
    f"{total_meses}"
)


# ============================================================
# 3. GASTO MENSUAL POR CATEGORÍA
# ============================================================

gasto_categoria_mes = (
    df_base_presupuesto
    .groupby(
        [
            "mes",
            "categoria_presupuesto"
        ]
    )
    ["cargo"]
    .sum()
    .unstack(
        fill_value=0
    )
)


# ============================================================
# 4. ASEGURAR QUE EXISTAN TODOS LOS MESES
# ============================================================

gasto_categoria_mes = (
    gasto_categoria_mes
    .reindex(
        todos_los_meses,
        fill_value=0
    )
)


# ============================================================
# 5. FUNCIÓN PARA ANALIZAR CADA CATEGORÍA
# ============================================================

def analizar_categoria(
    categoria,
    serie
):

    serie = (
        pd.to_numeric(
            serie,
            errors="coerce"
        )
        .fillna(0)
    )


    # --------------------------------------------------------
    # FRECUENCIA
    # --------------------------------------------------------

    meses_con_gasto = int(
        (serie > 0).sum()
    )

    frecuencia = (
        meses_con_gasto
        / total_meses
        if total_meses > 0
        else 0
    )


    # --------------------------------------------------------
    # VALORES HISTÓRICOS
    # --------------------------------------------------------

    gastos_positivos = (
        serie.loc[
            serie > 0
        ]
    )


    if len(gastos_positivos) > 0:

        mediana_activa = (
            gastos_positivos.median()
        )

        promedio_activo = (
            gastos_positivos.mean()
        )

        percentil_75 = (
            gastos_positivos.quantile(0.75)
        )

    else:

        mediana_activa = 0
        promedio_activo = 0
        percentil_75 = 0


    # Promedio distribuido entre TODOS los meses.
    #
    # Es especialmente útil para gastos periódicos:
    #
    # ejemplo:
    # seguro anual de $12,000
    # -> provisión mensual aproximada de $1,000

    provision_mensual = (
        serie.sum()
        / total_meses
        if total_meses > 0
        else 0
    )


    # --------------------------------------------------------
    # COMPORTAMIENTO RECIENTE
    # --------------------------------------------------------

    ultimos_3 = (
        serie
        .tail(3)
    )

    ultimos_6 = (
        serie
        .tail(6)
    )

    promedio_3m = (
        ultimos_3.mean()
        if len(ultimos_3) > 0
        else 0
    )

    mediana_6m = (
        ultimos_6.median()
        if len(ultimos_6) > 0
        else 0
    )


    # --------------------------------------------------------
    # TIPO DE COMPORTAMIENTO
    # --------------------------------------------------------

    if frecuencia >= 0.75:

        patron = "recurrente"

    elif frecuencia >= 0.25:

        patron = "periodico"

    else:

        patron = "ocasional"


    # --------------------------------------------------------
    # PRESUPUESTO SUGERIDO
    # --------------------------------------------------------

    if patron == "recurrente":

        # Para gastos frecuentes:
        #
        # usamos el comportamiento típico,
        # pero evitamos quedar por debajo
        # del comportamiento reciente.

        presupuesto = max(
            mediana_activa,
            mediana_6m
        )


    else:

        # Para gastos no mensuales:
        #
        # convertimos el gasto histórico
        # en una provisión mensual.

        presupuesto = provision_mensual


    # --------------------------------------------------------
    # REDONDEAR A $50
    # --------------------------------------------------------

    if presupuesto > 0:

        presupuesto = (
            np.ceil(
                presupuesto / 50
            )
            * 50
        )


    # --------------------------------------------------------
    # CLASIFICAR EL PAPEL EN EL PRESUPUESTO
    # --------------------------------------------------------

    if categoria == "ahorro_inversion":

        grupo = "ahorro"

    elif categoria == "pago_tdc":

        grupo = "salida_financiera"

    elif categoria in [
        "efectivo",
        "otros"
    ]:

        grupo = "requiere_revision"

    else:

        grupo = "gasto"


    return {
        "categoria":
            categoria,

        "grupo":
            grupo,

        "patron":
            patron,

        "meses_con_gasto":
            meses_con_gasto,

        "meses_analizados":
            total_meses,

        "frecuencia":
            frecuencia,

        "mediana_mes_activo":
            mediana_activa,

        "promedio_mes_activo":
            promedio_activo,

        "percentil_75":
            percentil_75,

        "promedio_ultimos_3m":
            promedio_3m,

        "mediana_ultimos_6m":
            mediana_6m,

        "provision_mensual":
            provision_mensual,

        "presupuesto_sugerido":
            presupuesto,
    }


# ============================================================
# 6. ANALIZAR TODAS LAS CATEGORÍAS
# ============================================================

registros_presupuesto = []

for categoria in gasto_categoria_mes.columns:

    registro = analizar_categoria(
        categoria,
        gasto_categoria_mes[categoria]
    )

    registros_presupuesto.append(
        registro
    )


df_presupuesto_sugerido = pd.DataFrame(
    registros_presupuesto
)


# ============================================================
# 7. ORDENAR
# ============================================================

orden_grupo = {
    "gasto": 1,
    "salida_financiera": 2,
    "ahorro": 3,
    "requiere_revision": 4,
}


df_presupuesto_sugerido[
    "orden_grupo"
] = (
    df_presupuesto_sugerido[
        "grupo"
    ]
    .map(orden_grupo)
    .fillna(99)
)


df_presupuesto_sugerido = (
    df_presupuesto_sugerido
    .sort_values(
        [
            "orden_grupo",
            "presupuesto_sugerido"
        ],
        ascending=[
            True,
            False
        ]
    )
    .drop(
        columns="orden_grupo"
    )
    .reset_index(drop=True)
)


# ============================================================
# 8. FORMATO DE FRECUENCIA
# ============================================================

df_presupuesto_sugerido[
    "frecuencia_pct"
] = (
    df_presupuesto_sugerido[
        "frecuencia"
    ]
    * 100
)


# ============================================================
# 9. PRESUPUESTO TOTAL
# ============================================================

presupuesto_total = (
    df_presupuesto_sugerido[
        "presupuesto_sugerido"
    ]
    .sum()
)


presupuesto_gasto = (
    df_presupuesto_sugerido.loc[
        df_presupuesto_sugerido[
            "grupo"
        ] == "gasto",
        "presupuesto_sugerido"
    ]
    .sum()
)


presupuesto_pago_tdc = (
    df_presupuesto_sugerido.loc[
        df_presupuesto_sugerido[
            "grupo"
        ] == "salida_financiera",
        "presupuesto_sugerido"
    ]
    .sum()
)


presupuesto_ahorro = (
    df_presupuesto_sugerido.loc[
        df_presupuesto_sugerido[
            "grupo"
        ] == "ahorro",
        "presupuesto_sugerido"
    ]
    .sum()
)


presupuesto_revision = (
    df_presupuesto_sugerido.loc[
        df_presupuesto_sugerido[
            "grupo"
        ] == "requiere_revision",
        "presupuesto_sugerido"
    ]
    .sum()
)


# ============================================================
# 10. MOSTRAR TABLA
# ============================================================

columnas_mostrar = [
    "categoria",
    "grupo",
    "patron",
    "meses_con_gasto",
    "frecuencia_pct",
    "mediana_mes_activo",
    "promedio_ultimos_3m",
    "provision_mensual",
    "presupuesto_sugerido",
]


print()
print("Presupuesto mensual sugerido")
print("-" * 70)

display(
    df_presupuesto_sugerido[
        columnas_mostrar
    ]
)


# ============================================================
# 11. RESUMEN
# ============================================================

print()
print("Resumen")
print("-" * 70)

print(
    f"Gastos identificados: "
    f"${presupuesto_gasto:,.2f}"
)

print(
    f"Pago TDC: "
    f"${presupuesto_pago_tdc:,.2f}"
)

print(
    f"Ahorro / inversión: "
    f"${presupuesto_ahorro:,.2f}"
)

print(
    f"Rubros por revisar: "
    f"${presupuesto_revision:,.2f}"
)

print("-" * 70)

print(
    f"Presupuesto mensual total sugerido: "
    f"${presupuesto_total:,.2f}"
)
```

```python
# ============================================================
# PRESENTACIÓN DEL PRESUPUESTO MENSUAL
# ============================================================

# Crear una vista sin modificar el DataFrame original
df_presupuesto_vista = (
    df_presupuesto_sugerido[
        [
            "categoria",
            "grupo",
            "patron",
            "meses_con_gasto",
            "frecuencia_pct",
            "mediana_mes_activo",
            "promedio_ultimos_3m",
            "provision_mensual",
            "presupuesto_sugerido",
        ]
    ]
    .copy()
)


# ============================================================
# 1. NOMBRES MÁS LEGIBLES
# ============================================================

df_presupuesto_vista = df_presupuesto_vista.rename(
    columns={
        "categoria": "Categoría",
        "grupo": "Grupo",
        "patron": "Patrón",
        "meses_con_gasto": "Meses con gasto",
        "frecuencia_pct": "Frecuencia",
        "mediana_mes_activo": "Mediana mensual",
        "promedio_ultimos_3m": "Promedio 3 meses",
        "provision_mensual": "Provisión mensual",
        "presupuesto_sugerido": "Presupuesto sugerido",
    }
)


# ============================================================
# 2. HACER LEGIBLES LOS VALORES DE TEXTO
# ============================================================

def formato_texto(valor):
    return (
        str(valor)
        .replace("_", " ")
        .strip()
        .title()
    )


for columna in [
    "Categoría",
    "Grupo",
    "Patrón",
]:
    df_presupuesto_vista[columna] = (
        df_presupuesto_vista[columna]
        .apply(formato_texto)
    )


# ============================================================
# 3. ESTILOS POR GRUPO
# ============================================================

def colorear_grupo(fila):

    grupo = fila["Grupo"]

    if grupo == "Gasto":
        estilo = (
            "background-color: #fce8e6;"
            "color: #202124;"
        )

    elif grupo == "Salida Financiera":
        estilo = (
            "background-color: #fff4cc;"
            "color: #202124;"
        )

    elif grupo == "Ahorro":
        estilo = (
            "background-color: #e6f4ea;"
            "color: #202124;"
        )

    elif grupo == "Requiere Revision":
        estilo = (
            "background-color: #f3e8fd;"
            "color: #202124;"
        )

    else:
        estilo = "color: #202124;"

    return [estilo] * len(fila)

# ============================================================
# 4. CONSTRUIR TABLA ESTILIZADA
# ============================================================

tabla_presupuesto = (
    df_presupuesto_vista
    .style

    # Ocultar índice
    .hide(axis="index")

    # Formato numérico
    .format(
        {
            "Meses con gasto": "{:,.0f}",
            "Frecuencia": "{:.0f}%",
            "Mediana mensual": "${:,.0f}",
            "Promedio 3 meses": "${:,.0f}",
            "Provisión mensual": "${:,.0f}",
            "Presupuesto sugerido": "${:,.0f}",
        },
        na_rep="-"
    )

    # Color según el tipo de rubro
    .apply(
        colorear_grupo,
        axis=1
    )

    # Resaltar la columna más importante
    .background_gradient(
        subset=["Presupuesto sugerido"],
        cmap="Blues"
    )

    # Estilo general
    .set_properties(
        **{
            "font-size": "12px",
            "padding": "7px 10px",
            "border-bottom": "1px solid #dddddd",
        }
    )

    # Alinear números a la derecha
    .set_properties(
        subset=[
            "Meses con gasto",
            "Frecuencia",
            "Mediana mensual",
            "Promedio 3 meses",
            "Provisión mensual",
            "Presupuesto sugerido",
        ],
        **{
            "text-align": "right"
        }
    )

    # Encabezados
    .set_table_styles(
        [
            {
                "selector": "th",
                "props": [
                    ("background-color", "#263238"),
                    ("color", "white"),
                    ("font-weight", "bold"),
                    ("text-align", "center"),
                    ("padding", "8px"),
                ],
            },
            {
                "selector": "caption",
                "props": [
                    ("caption-side", "top"),
                    ("font-size", "18px"),
                    ("font-weight", "bold"),
                    ("text-align", "left"),
                    ("padding", "8px 0 12px 0"),
                ],
            },
        ]
    )

    .set_caption(
        "Presupuesto mensual sugerido"
    )
)


display(tabla_presupuesto)
```
